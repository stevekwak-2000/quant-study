```python
import ccxt
import pandas as pd

ex = ccxt.upbit()
ohlcv = ex.fetch_ohlcv('BTC/KRW', '1d', limit=200)

df = pd.DataFrame(ohlcv, columns=['ts','open','high','low','close','volume'])
df['ts'] = pd.to_datetime(df['ts'], unit='ms')
df = df.set_index('ts')

df['close'].plot(title='BTC/KRW')
```
